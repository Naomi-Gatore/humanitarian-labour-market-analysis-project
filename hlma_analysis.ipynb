```python
# Check rows where both date_posted and deadline columns have missing values
missing_both = df[df["date_posted"].isna() & df["deadline"].isna()]
missing_both["organization"].value_counts()
```

```python
# Check for missing date_posted values in DevNet Jobs
devnetjobs_rows = df[df["organization"] != "Amref"]
devnetjobs_rows = devnetjobs_rows[~devnetjobs_rows["organization"].isin(["MSF USA", "GiveDirectly", "Human Rights Watch"])]

print(f"Total non-ATS (DevNetJobs) rows: {len(devnetjobs_rows)}")
print(f"Missing date_posted in DevNetJobs: {devnetjobs_rows['date_posted'].isna().sum()}")
print(f"Missing deadline in DevNetJobs: {devnetjobs_rows['deadline'].isna().sum()}")
```

```python
# Check where specific data source the 6 duplicate organisations fixed by fuzzy matching are from
duplicate_org_names = [
    "Care", "CARE", "Goal", "GOAL", "Searca", "SEARCA",
    "The Better Cotton Initiative", "Better Cotton Initiative",
    "Asian Development Bank (ADB)", "Asian Development Bank",
    "United Nations University (UNU)", "United Nations University"
]

check_source = df[df["organization"].isin(duplicate_org_names)]
check_source[["organization", "source_url"]].drop_duplicates()
```

```python
# Check if hybrid and roving work_arrangements exist in the data set
# Hybrid only exists in the location column at the moment
hybrid_check = df[df["location"].str.contains("hybrid", case=False, na=False)]
roving_check = df[df["location"].str.contains("roving", case=False, na=False)]

print(f"Rows mentioning 'Hybrid': {len(hybrid_check)}")
print(f"Rows mentioning 'Roving': {len(roving_check)}")

print("\n=== Hybrid matches ===")
print(hybrid_check[["organization", "location"]].to_string())

print("\n=== Roving matches ===")
print(roving_check[["organization", "location"]].to_string())
```
